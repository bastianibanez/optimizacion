```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex", "coin"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile 10_3_11.mod
param a1;  param b1; param c1;     # 36, 9, 6
param a2;  param b2;               # 36, 3
param RHS;                         # 3  (x1+x2 <= RHS)

var x1 integer >= 0;
var x2 integer >= 0;

maximize Z:
      a1*x1 + b1*x1^2 - c1*x1^3
    + a2*x2 - b2*x2^2 ;

s.t. Cap:  x1 + x2 <= RHS;
```

```python
%%writefile 10_3_11.dat
param a1 := 36;
param b1 :=  9;
param c1 :=  6;
param a2 := 36;
param b2 :=  3;
param RHS := 3;
```

```python
ampl.read('10_3_11.mod')
ampl.readData('10_3_11.dat')

ampl.option['solver']     = 'couenne'
ampl.option['solver_msg'] = 0      # sin log largo
ampl.solve()
ampl.display('x1','x2','Z')
```
