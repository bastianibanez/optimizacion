```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex", "coin"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile 10_3_18.mod
param U;                       # cota superior de x3  ( =3 )

var x1 >= 0;
var x2 >= 0;
var x3 >= 0 <= U;              # 0 <= x3 <= U
var y  binary;                 # 1 si x3 > 0

maximize Z:
      3*x1 + 7*x2 + 6*(-y + x3);      # f(x3) = -y + x3

s.t. Cap1:  x1 + 3*x2 + 2*x3 <= 6;
s.t. Cap2:  x1 +   x2        <= 5;

# Vinculo "si x3>0 entonces y=1"
s.t. Link:  x3 <= U * y;
```

```python
%%writefile 10_3_18.dat
param U := 3 ;
```

```python
ampl.read('10_3_18.mod')
ampl.readData('10_3_18.dat')

ampl.option['solver'] = 'highs'      # CBC también sirve
ampl.solve()
ampl.display('x1','x2','x3','y','Z')
```
