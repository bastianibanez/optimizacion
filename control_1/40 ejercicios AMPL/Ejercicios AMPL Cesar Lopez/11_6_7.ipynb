```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile assignment.mod
set I;                      # trabajadores
set J;                      # tareas
param c{I,J};               # costo de asignar i a j

var x{I,J} binary;          # 1 si i realiza j

minimize Coste:
    sum{(i,j) in I cross J} c[i,j] * x[i,j];

s.t. UnTrabajadorPorTarea{j in J}:
    sum{i in I} x[i,j] = 1;

s.t. UnaTareaPorTrabajador{i in I}:
    sum{j in J} x[i,j] = 1;
```

```python
%%writefile assignment.dat
set I := 1 2 3 4 5 ;
set J := 1 2 3 4 5 ;

param c :  1   2   3   4   5 :=
 1   39  65  69  66  57
 2   64  84  24  92  22
 3   49  50  61  31  45
 4   48  45  55  23  50
 5   59  34  30  34  18 ;
```

```python
ampl.read('assignment.mod')
ampl.readData('assignment.dat')

ampl.option['solver']      = 'cbc'          # o 'highs'
ampl.option['solver_msg']  = 1              # imprimir árbol B&B
ampl.option['cbc_options'] = 'logLevel 1'   # nodos / ramas / cortes

ampl.solve()
ampl.display('Coste')
ampl.display('x')
```
