```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile 8_3_4.mod
set SWIM;            # nadadores  (5)
set STROKE;          # estilos    (4)

param t{SWIM,STROKE};   # tiempo (costo)

# x[i,j] = 1 si el nadador i nada el estilo j
var x{SWIM,STROKE} binary;

minimize TotalTime: sum{i in SWIM, j in STROKE} t[i,j]*x[i,j];

# exactamente 1 nadador por estilo
s.t. OnePerStroke{j in STROKE}: sum{i in SWIM} x[i,j] = 1;

# a lo mas 1 estilo por nadador  (queda 1 sin asignar)
s.t. MaxOneStyle{i in SWIM}:    sum{j in STROKE} x[i,j] <= 1;
```

```python
%%writefile 8_3_4.dat
# -------- 8_3_4.dat -------------------------------------
set SWIM   := Carl Chris David Tony Ken ;
set STROKE := Dorso Pecho Mariposa Libre ;

param t :
          Dorso Pecho Mariposa Libre :=
  Carl    37.7  43.4   33.3     29.2
  Chris   32.9  33.1   28.5     26.4
  David   33.8  42.2   38.9     29.6
  Tony    37.0  34.7   30.4     28.5
  Ken     35.4  41.8   33.6     31.1 ;
```

```python
# 3) leer, resolver y mostrar
ampl.read('8_3_4.mod')
ampl.readData('8_3_4.dat')
ampl.option['solver'] = 'highs'
ampl.solve()
ampl.display('x','TotalTime')
```
