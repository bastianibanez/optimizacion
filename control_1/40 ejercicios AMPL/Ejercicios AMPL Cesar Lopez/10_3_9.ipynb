```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex", "coin"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile 10_3_9.mod
param a1;      #  coeficientes  3  ( x1^2 )
param b1;      #  coeficientes  1  ( x1^3 )
param a2;      #               5  ( x2^2 )
param b2;      #               1  ( x2^3 )
param RHS;     # lado derecho de x1 + 2 x2 <= RHS (=4)

var x1 integer >= 0;          # x1 ∈ Z+
var x2 integer >= 0;          # x2 ∈ Z+

maximize Z:
      a1 * x1^2  -  b1 * x1^3
    + a2 * x2^2  -  b2 * x2^3 ;

s.t. Cap:  x1 + 2 * x2  <= RHS;
```

```python
%%writefile 10_3_9.dat
param a1  := 3 ;
param b1  := 1 ;
param a2  := 5 ;
param b2  := 1 ;
param RHS := 4 ;
```

```python
ampl.read('10_3_9.mod')
ampl.readData('10_3_9.dat')

ampl.option['solver']     = 'couenne'
ampl.option['solver_msg'] = 0      # (silencio)
ampl.solve()
ampl.display('x1','x2','Z')
```
