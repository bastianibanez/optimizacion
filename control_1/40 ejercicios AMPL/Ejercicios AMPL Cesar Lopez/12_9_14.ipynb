```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex", "coin"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile 12_9_14.mod
param r   > 0;                      # parametro de barrera

var x1 >= 3;
var x2 >= 3;

maximize P:
    -2*x1 - (x2-3)^2
    - r/(x1-3) - r/(x2-3);          # f(x) + barrera
```

```python
%%writefile 12_9_14.dat
param r := 1 ;      # cambia a 0.01, 1e-4, 1e-6
```

```python
for r in [1, 1e-2, 1e-4, 1e-6]:
    ampl.reset()                          # limpia todo
    ampl.read('12_9_14.mod')              # lee el modelo
    ampl.param['r'] = r                   # asigna valor al parámetro
    ampl.option['solver'] = 'ipopt'       # NL solver continuo
    ampl.solve()

    print(f"r = {r:g}")
    ampl.display('x1','x2','P')
    print('-'*30)
```
