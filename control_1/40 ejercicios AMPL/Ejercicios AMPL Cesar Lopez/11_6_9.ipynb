```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex", "coin"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile modelo_11_6_9.mod
set J; set PAIRS within {J,J};
param c{J}; param d{J};
var x{J} binary; var y{PAIRS} binary;
maximize Z:
      sum{j in J} c[j]*x[j]
    - sum{j in J} d[j]^2 * x[j]
    - 2*sum{(i,j) in PAIRS} d[i]*d[j]*y[i,j];
s.t. Link1{(i,j) in PAIRS}: y[i,j] <= x[i];
s.t. Link2{(i,j) in PAIRS}: y[i,j] <= x[j];
s.t. Link3{(i,j) in PAIRS}: y[i,j] >= x[i]+x[j]-1;
```

```python
%%writefile datos_11_6_9.dat
set J := 1 2 3 4;
set PAIRS := (1,2) (1,3) (1,4) (2,3) (2,4) (3,4);
param c := 1 80  2 60  3 40  4 20 ;
param d := 1  7  2  5  3  3  4  2 ;
end;
```

```python
ampl = AMPL()
ampl.read('modelo_11_6_9.mod')
ampl.readData('datos_11_6_9.dat')
ampl.option['solver'] = 'highs'
ampl.solve()
ampl.display('x','Z')

```
