```python
# 1. Instalación e inicialización de AMPL con tu licencia Community Edition
from amplpy import ampl_notebook

# Inicializa AMPL en Colab con tu UUID (Community Edition) Inicializar AMPL
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex"],  # solvers disponibles
    license_uuid="936b618d-a013-406f-9809-49679f557c26"
)
```

```python
%%writefile 12_8_8.mod
set K;                      # indices de los puntos de quiebre

param b{K};                 # valor de x1 en el punto k
param y{K};                 # valor de x1^2  (= b[k]^2)

var lambda{K} >= 0;         # ponderaciones convexas
s.t. SumLam: sum{k in K} lambda[k] = 1;

var x1_hat;                 # valor aproximado de x1
var x1sq_hat;               # valor aproximado de x1^2
var x2 >= 0;

# definiciones lineales (convex comb.)
s.t. DefX1:    x1_hat   = sum{k in K} b[k]*lambda[k];
s.t. DefX1sq:  x1sq_hat = sum{k in K} y[k]*lambda[k];

maximize Z: 5*x1_hat + x2;

s.t. R1:  2*x1sq_hat + x2 <= 13;
s.t. R2:     x1sq_hat + x2 <=  9;
```

```python
%%writefile 12_8_8.dat
# 4 puntos de quiebre: 0,1,2,3
set K := 0 1 2 3 ;

param b :=
 0 0
 1 1
 2 2
 3 3 ;

param y :=
 0 0
 1 1
 2 4
 3 9 ;
```

```python
ampl.read('12_8_8.mod')
ampl.readData('12_8_8.dat')

ampl.option['solver'] = 'highs'   # CBC también sirve
ampl.solve()

ampl.display('x1_hat','x1sq_hat','x2','Z','lambda')
```
