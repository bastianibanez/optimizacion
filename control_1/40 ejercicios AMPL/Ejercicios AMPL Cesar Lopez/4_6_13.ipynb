```python
%%writefile 4_6_13.mod
# Variables no negativas
var x1p >= 0;  # x1 positivo
var x1m >= 0;  # x1 negativo
var x2p >= 0;  # x2 desplazado (x2 + 3)

# Funcion objetivo (se ignora la constante -12)
maximize Z: -x1p + x1m + 4*x2p;

# Restricciones
s.t. R1: -3*x1p + 3*x1m +   x2p <= 9;
s.t. R2:  x1p  -  x1m + 2*x2p <= 10;
```

```python
%%writefile 4_6_13.dat
# No se requieren parámetros externos.
```

```python
from amplpy import AMPL

ampl = AMPL()
ampl.read("4_6_13.mod")
ampl.readData("4_6_13.dat")

ampl.option['solver'] = 'highs'
ampl.option['solver_msg'] = 1
ampl.option['highs_options'] = 'presolve=off simplex_strategy=1'

ampl.solve()
ampl.display('x1p', 'x1m', 'x2p', 'Z')
```
